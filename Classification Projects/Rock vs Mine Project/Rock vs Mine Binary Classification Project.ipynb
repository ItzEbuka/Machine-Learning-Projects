### Import the necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
```

### Load the dataset

```python
df = pd.read_csv('copy of sonar data.csv', header = None)
```

### Data Preprocessing

#### Checking for missing values

```python
df.isnull().sum()
```

There is no missing value.

### Splitting the data

```python
# Features
X = df.drop(columns= 60, axis = 1).values

# Target
y = df[60].values
```

```python
print(X)
```

```python
print(y)
```

```python
# Getting the training and testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.1, random_state = 42, stratify = y)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### Building the Logistic Regression Model

```python
# Create an instance of the model
log_model = LogisticRegression()
```

```python
# Fit the training data
log_model.fit(X_train, y_train)
```

```python
training_pred = log_model.predict(X_train)
```

```python
print('Training Accuracy: ', accuracy_score(y_train, training_pred))
```

The training accuracy is approx. 82.35%

```python
# Using the testing set
testing_pred = log_model.predict(X_test)
print('Testing Accuracy: ', accuracy_score(y_test, testing_pred))
print('\n')
print('Confusion Matrix: ', confusion_matrix(y_test, testing_pred))
```

The testing accuracy is approximately 85.71%. Also, from the confusion matrix, only about 3 points were not classified correctly.
