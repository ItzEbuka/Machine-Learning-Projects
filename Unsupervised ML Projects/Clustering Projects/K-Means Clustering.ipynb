```python
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

> For an Unspuervised ML algorithm, we are not predicting outcomes rather, we seek to find patterns in the dataset 

```python
from sklearn.datasets import make_blobs

data = make_blobs(n_samples=200, n_features = 2, centers = 4, cluster_std = 1.8, random_state = 101)
```

```python
# Checking out the artificially created dataset

data[0].shape
```

```python
# Looking at some of the data visually
plt.scatter(data[0][:,0], data[0][:,1], c = data[1])
```

```python
from sklearn.cluster import KMeans
```

```python
# Create an instance of the model
kmeans = KMeans(n_clusters = 4)
```

```python
# Fit the model
kmeans.fit(data[0])
```

```python
# Lets check out some things in the model
kmeans.cluster_centers_
```

```python
# What was predicted by the algorithm?
kmeans.labels_
```

```python
# Actual values in the dataset

data[1]
```

```python
# Lets visualize the results
fig, (ax1, ax2)  = plt.subplots(1, 2, sharey = True, figsize = (10, 7))

ax1.set_title("K-means")
ax1.scatter(data[0][:,0], data[0][:,1], c = kmeans.labels_, cmap = 'rainbow')
ax2.set_title("Originsl data")
ax2.scatter(data[0][:,0], data[0][:,1], c = data[1], cmap = 'rainbow')
```
